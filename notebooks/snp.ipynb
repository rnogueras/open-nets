# Single Neuron Perceptron

```python
from open_nets.layers import Linear
from open_nets.activation import Sigmoid
from open_nets.losses import MSELoss
```

```python
# Dataset
x = 1
y = 0.5

# Model
linear = Linear()
sigmoid = Sigmoid()
mse = MSELoss()

# Optimizer
learning_rate = 2.5

epoch = 0
```

```python
# Epoch start
print(f"\nEpoch {epoch}")

# Forward propagation
z = linear(x)
y_pred = sigmoid(z)
print(f"Prediction: {y_pred}")

# Loss computation
loss = mse(y_pred, y)
print(f"Loss: {loss}")

# Back propagation
grad_loss = mse.backward()
grad_s = sigmoid.backward(grad_loss)
grad_w, grad_b, grad_x = linear.backward(grad_s)

# Parameter update
linear.w = linear.w - learning_rate * grad_w
linear.b = linear.b - learning_rate * grad_b

# Epoch update
epoch += 1
```
